```python
# from icd.core import *
from icd.interface import *

import ipywidgets as widgets
from IPython.display import display


logoOcra = widgets.Image(
    value=open('bin/infn-ocra.png', 'rb').read(),
    format='png',
    width=150
)

logoUnige = widgets.Image(
    value=open('bin/unige.jpg', 'rb').read(),
    format='jpg',
    width=150
)

pageTitle = widgets.HTML(
    '''<h1>International Cosmic Day</h1>
    <h2>Data analysis application</h2>
    '''
)

header = widgets.HBox(
    [
        widgets.Box([pageTitle], layout=widgets.Layout(
            width='100%', display='flex', #justify_content='center'
        )),
        logoUnige,
        logoOcra
    ],
    layout=widgets.Layout(align_items='center', margin='50px 0px')
)

mode_selector = widgets.ToggleButtons(
    options=[
        ('Count Analysis', 'count'),
        ('Angle Analysis', 'angular')
    ],
    description='Mode',
    button_style='',
    # icons=['area-chart', 'line-chart']
)

# -------------------
# LEFT PANEL (APP)
# -------------------

# Angular UI
angle_input = widgets.FloatText(description="Angle:")
intensity_input = widgets.FloatText(description="Intensity:")

angular_ui = widgets.VBox([
    widgets.HTML("<h3>Angular Dependence Analysis</h3>"),
    angle_input,
    intensity_input
])

# Count UI
count_input = widgets.IntText(description="Counts:")
time_input = widgets.FloatText(description="Time (s):")

count_ui = widgets.VBox([
    widgets.HTML("<h3>Count Analysis</h3>"),
    count_input,
    time_input
])

left_panel = widgets.VBox()
left_panel.layout = widgets.Layout(width="70%", padding="10px")

# -------------------
# RIGHT PANEL (TEXT)
# -------------------

right_panel = widgets.HTML()
right_panel.layout = widgets.Layout(
    width="30%",
    padding="20px",
    border_left="2px solid lightgray",
    overflow="auto",
    height="500px"
)

# -------------------
# DYNAMIC UPDATE
# -------------------

def update_ui(change=None):
    # Update left panel
    if mode_selector.value == "angular":
        left_panel.children = [angular_ui]
        right_panel.value = """
        <h3>Angular Dependence</h3>
        <p>
        Study how the cosmic ray rate depends on the zenith angle.
        </p>
        <ul>
            <li>Insert angle in degrees</li>
            <li>Insert measured intensity</li>
        </ul>
        """
    else:
        left_panel.children = [count_ui]
        right_panel.value = """
        <h3>Count Analysis</h3>
        <p>
        Analyze counting rates and estimate uncertainties.
        </p>
        <ul>
            <li>Insert number of counts</li>
            <li>Insert acquisition time</li>
        </ul>
        """

mode_selector.observe(update_ui, names="value")

# -------------------
# RUN BUTTON + OUTPUT
# -------------------

run_button = widgets.Button(
    description="Run Analysis",
    button_style="success"
)

output = widgets.Output()

def run_analysis(b):
    output.clear_output()
    with output:
        if mode_selector.value == "angular":
            print("Running angular analysis...")
            print(angle_input.value, intensity_input.value)
        else:
            print("Running count analysis...")
            print(count_input.value, time_input.value)

run_button.on_click(run_analysis)

# Add button/output to left panel
left_panel.children = [angular_ui, run_button, output]

# -------------------
# MAIN LAYOUT
# -------------------

main_area = widgets.HBox([left_panel, right_panel])

app = widgets.VBox([
    header,
    mode_selector,
    main_area
])

# Initialize
update_ui()

display(app)
```
